```python
# Load the dataset  //  Task -1
from sklearn.datasets import load_breast_cancer
data = load_breast_cancer()
x = data.data
y = data.target
```

```python
# Explore the dataset  //  Task -2

# Display the shape of X and y.
print("Shape of X:", x.shape)
print("Shape of y:", y.shape)

```

```python
# 2. Print the first five rows.
print("First five rows of X:")
print(x[:5])
```

```python
 # 3. Identify:
# Number of samples
print("Number of samples:", x.shape[0])
# Number of features
print("Number of features:", x.shape[1])
# Number of classes
print("Number of classes:", len(np.unique(y)))
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=42
)

print("Training:", x_train.shape)
print("Testing :", x_test.shape)
```

```python
# Standardize the features
from sklearn.preprocessing import StandardScaler

# Create the scaler
scaler = StandardScaler()

# Fit on training data and transform both datasets
x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)

print("Feature scaling completed successfully.")
print("Training data shape:", x_train.shape)
print("Testing data shape :", x_test.shape)
```

```python
from sklearn.linear_model import LogisticRegression

# Create the Logistic Regression model with default parameters
model = LogisticRegression()

# Train the model
model.fit(x_train, y_train)

print("Logistic Regression model trained successfully.")
```

```python
# Task 6: Make Predictions
# Predict class labels for the test set
y_pred = model.predict(x_test)

print("Predicted Class Labels:")
print(y_pred)
```

```python
# Task 7: Evaluate the Model
# compute following paramerters to evaluate you model:
# Accuracy, Precision, Recall, F1 Score, Confusion Matrix

# Import evaluation metrics
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, classification_report

# Predict on test data
y_pred = model.predict(x_test)
```

```python
# 1. Accuracy
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)
```

```python
# 2. Precision
precision = precision_score(y_test, y_pred)
print("Precision:", precision)
```

```python
# 3. Recall
recall = recall_score(y_test, y_pred)
print("Recall:", recall)
```

```python
# 4. F1-Score
f1 = f1_score(y_test, y_pred)
print("F1-Score:", f1)
```

```python
# 5. Confusion Matrix
cm = confusion_matrix(y_test, y_pred)
print("\nConfusion Matrix:")
print(cm)
```

```python
# Task-8: Display the coefficients of the trained model
print("Model Coefficients:")
print(model.coef_)

print("\nIntercept:")
print(model.intercept_)
```

```python
# Answer and questions
import pandas as pd

# Create DataFrame
coefficients = pd.DataFrame({
    'Feature': data.feature_names,
    'Coefficient': model.coef_[0]
})

# Largest positive coefficient
largest_positive = coefficients.loc[coefficients['Coefficient'].idxmax()]

# Largest negative coefficient
largest_negative = coefficients.loc[coefficients['Coefficient'].idxmin()]

print("Largest Positive Coefficient:")
print(largest_positive)

print("\nLargest Negative Coefficient:")
print(largest_negative)
```
