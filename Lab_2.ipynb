```python
# Part 1: Working with Dictionary
```

```python
# A data structure used to store data in key-value pairs. 
# It is a collection of unordered, mutable, and indexed data. Each key is unique and maps to a value.
Student = {
    "Name": ("Subhash Kumar"),
    "Age": 23,
    "faculty": "BCE"
}   
print(Student)
```

```python
#creating a dictionary
Student = {
    "Name": ["Subhash", " Kumar"],
    "Age": 23,
    "faculty": "BCE"
}   
print(Student)
```

```python
#Accessing values from the dictionary
print(Student["Name"])
print(Student["Age"])
print(Student["faculty"])

print(Student.get("Name")[0])
```

```python
#use get() method to acess values from the dictionary
print(Student.get("Name"))
print(Student.get("Age"))
print(Student.get("faculty"))
```

```python
#Adding a new key-value pair to the dictionary
Student["Department"] = "Computer Science"
print(Student)
```

```python
# update value of existing key
Student["Age"] = 24  # update age from 23 to 24
print(Student["Age"])
print(Student)
```

```python
# Remove items from the dictionary
del Student["faculty"]  # remove the key "faculty" and its value
print(Student)
```

```python
# Remove items from the dictionary using pop() method
Student.pop("Department")  
print(Student)
```

```python
# Remove last items from the dictionary
Student.popitem() 
print(Student)
```

```python
# Clear Dictionary
Student.clear()
print(Student)
```

```python
# Copying a dictionary

# Create a copy of the dictionary
Student_copy = Student.copy()
print(Student_copy)
```

```python
# Get All Keys from the dictionary
print(Student_copy.keys())

# Get All Values from the dictionary
print(Student_copy.values())

# Get All Items from the dictionary
print(Student_copy.items())
```

```python
#Check Keys Exits in the dictionary
print("Name" in Student) 
print("Age" in Student)
```

```python
# get the length of the dictionary
print(len(Student))
```

```python
#Loop through the dictionary and values
for key in Student:
    print(key, ":", Student[key])
```

```python
#Loopthrough items in the dictionary
for key, value in Student.items():
    print(key, ":", value)
```

```python
print(Student.items())
```

```python
#creating a data frame containing student details such as Name, Age, and Faculty
import pandas as pd
Data = {
    "Name": ["Subhash", "Kumar", "Ravi", "Anil"],
    "Age": [20, 22, 21, 23],
    "Faculty": ["Engineering", "Medicine", "Law", "Business"]
}
df = pd.DataFrame(Data)

# Add Serial Number
df.insert(0, "S.N", range(1, len(df) + 1))

# Print without default index
print(df.to_string(index=False))
```

```python
# Display the first 3 rows of the data frame
print(df.head(3))

# Display the last 2 rows of the data frame
print(df.tail(2))
```

```python
# Find the total number of rows and columns in the data frame
print(df.shape)
```

```python
#Display all coulumns in the data frame
print(df.columns)
```

```python
df['Marks'] = df['Marks'] + 5  #Adding 5 marks to each value in the "Marks" column of the data frame
print(df)
```

```python
# Remove the first row without modifying the original DataFrame
df_no_first = df.drop(index=0)
print("Without first row:\n", df_no_first)

# Sort by Marks ascending
df_sorted_asc = df.sort_values(by='Marks', ascending=True).reset_index(drop=True)
print("\nSorted by Marks (asc):\n", df_sorted_asc)

# Sort by Marks descending
df_sorted_desc = df.sort_values(by='Marks', ascending=False).reset_index(drop=True)
print("\nSorted by Marks (desc):\n", df_sorted_desc)

# Concatenate another DataFrame with matching columns
df2 = pd.DataFrame({
    'S.N': [5],
    'Name': ['Neha'],
    'Age': [24],
    'Faculty': ['Science'],
    'Marks': [95],
    'Grade': ['A']
})

df_concat = pd.concat([df, df2], ignore_index=True)
print("\nConcatenated DataFrame:\n", df_concat)


```
